## Linear Regression Implementation Using NumPy


### Objective

The objective of this project is to implement a simple linear regression model from scratch using only NumPy.

The model learns the relationship between input and output data without using any machine learning libraries.

### problem Statement

In real-world scenarios, we often need to predict numerical values based on input features.

This project simulates how a machine learning model learns a linear relationship between variables.

### Dataset Overview

A simple synthetic dataset is used where:

- X represents input values (independent variable)

- y represents output values (dependent variable)

The relationship between X and y is approximately linear.

```python
import numpy as np

np.random.seed(42)

X = np.arange(1, 101)

noise = np.random.randn(100) * 2.0

y = 2.5 * X + 5 + noise

print("X (first 5):", X[:5])
print("y (first 5):", y[:5])
```

### Model Intuition

The model assumes a linear relationship:

- Output depends on input through a straight-line equation

- The goal is to find the best-fitting line for the data

- The model learns by adjusting slope and bias to reduce prediction error over time

### Model Parameters

The model learns two parameters:

- slope (controls steepness of the line)

- bias (controls vertical shift)

These parameters are adjusted during training to reduce error.


```python
m = 0.0

b = 0.0

learning_rate = 0.0001
iterations = 5000
n = len(X)
loss_history = []
```

### Untrained Model Prediction
The model generates predictions by applying the learned relationship to input data.

This step is repeated multiple times during training.

```python
y_pred_initial = m * X + b
print("Predicted y (Initial):", y_pred_initial)
```

### Loss Before Training

The loss function measures how far predictions are from actual values.

- A smaller loss means better predictions

- The goal of training is to minimize this loss


```python
error  = y - y_pred_initial

initial_loss = np.mean(error ** 2)
print("Initial Loss:", initial_loss)
```

### Training Process

The model learns by:

- making predictions

- calculating error

- adjusting parameters

- repeating this process multiple times

This iterative process helps the model improve gradually.


```python
for i in range(iterations):

    # prediction
    y_pred = m * X + b

    # error
    error = y - y_pred

    # loss
    loss = np.mean(error ** 2)
    loss_history.append(loss)

    # gradients
    dm = (-2/n) * np.sum(X * error)
    db = (-2/n) * np.sum(error)

    # update
    m -= learning_rate * dm
    b -= learning_rate * db

    # logging
    if i % 500 == 0:
        print(f"Iteration {i:5d}: Loss={loss:.4f}, m={m:.4f}, b={b:.4f}")
```

### Learning Intuition

Initially, the model starts with poor predictions because slope and bias are set to zero.

During training:

- prediction errors are calculated

- gradients determine how parameters should change

- slope and bias are updated repeatedly

- loss gradually decreases over time

As iterations increase, the model learns the best-fitting line for the dataset.

### Model Evaluation

After training, the learned parameters are used to generate final predictions.

The model is evaluated by:

- comparing predicted values with actual values

- observing reduction in loss

- checking whether the model successfully captured the linear relationship

A lower final loss indicates better model performance.

### Final Result

After training:

- the model learns optimized slope and bias values

- prediction accuracy improves significantly

- loss becomes much smaller compared to the initial stage

```python
final_pred = m * X + b
final_loss = np.mean((y - final_pred) ** 2)
loss_reduction = ((loss_history[0] - loss_history[-1]) / loss_history[0]) * 100
 
print("\n════════════════════════════════")
print("         FINAL RESULTS         ")
print("════════════════════════════════")
print(f"Learned Slope  (m) : {m:.4f}  (true value: 2.5)")
print(f"Learned Bias   (b) : {b:.4f}  (true value: 5.0)")
print(f"Initial Loss       : {loss_history[0]:.4f}")
print(f"Final Loss         : {loss_history[-1]:.4f}")
print(f"Loss Reduction     : {loss_reduction:.2f}%")
print("════════════════════════════════")
```
