### Random Module

Random module is used to generate random numbers for simulations and ML tasks.



```python
import numpy as np

np.random.seed(42)              # fixes randomness for reproducibility

print(np.random.rand(3))        # uniform random values [0,1)
print(np.random.randn(3))       # normal distribution (mean=0, std=1)
print(np.random.randint(1, 10, 5))  # random integers
print(np.random.choice([1,2,3,4], size=3))  # random selection
```

### Shuffle

Shuffles array elements in-place.

```python
arr = np.array([1,2,3,4,5])
np.random.shuffle(arr)
print(arr)
```

### Advanced Aggregations & Stats

Functions that summarize data into statistical measures.

## Median

Middle value of sorted data.


```python
arr = np.array([1, 2, 3, 100])
print(np.median(arr))
```

## Percentile

Value below which a percentage of data falls.

```python
arr = np.array([1, 2, 3, 4, 5])

print(np.percentile(arr, 50))  # median
print(np.percentile(arr, 25))  # lower quarter
print(np.percentile(arr, 75))  # upper quarter
```

# Correlation Coefficient

Measures how strongly two variables move together.

***Value Range***

- +1 → perfect positive correlation 

- 0 → no correlation  

- -1 → perfect negative correlation  


```python
x = np.array([1,2,3,4])
y = np.array([2,4,6,8])

print(np.corrcoef(x, y))
```

# Covariance

Measures how two variables change together.

- Positive → both increase together  

- Negative → one increases, other decreases  

- Around 0 → no clear relationship  

***Formula***

cov(X,Y) = Σ (xi - mean(X)) (yi - mean(Y)) / n


```python
x = np.array([1, 2, 3, 4])
y = np.array([8, 6, 4, 2])

cov = np.cov(x, y)

print(cov)
```

# 3. Sorting & Unique Operations

Used to organize and analyze data.

## Sorting

Arranges elements in ascending order.

```python
arr = np.array([3,1,5,2])

print(np.sort(arr))
```

# np.argsort( )

Returns the indices that would sort an array.


```python
arr = np.array([3,1,5,2])

print(np.argsort(arr))
```

***Meaning***

`[1, 3, 0, 2]` means:

- smallest value is at index 1  

- next at index 3  

- next at index 0  

- last at index 2  

***Using Indices***


```python
indices = np.argsort(arr)
print(arr[indices])
```

## Unique

Returns unique elements from array.


```python
arr = np.array([1,2,2,3,3,3])

print(np.unique(arr))
```

## Bincount

Counts frequency of each integer value.


```python
arr = np.array([1,2,2,3,3,3])

print(np.bincount(arr))
```

our array:

[1, 2, 2, 3, 3, 3]

Now count:

0 → 0 times

1 → 1 time

2 → 2 times

3 → 3 times

So result becomes:

[0, 1, 2, 3]
