## 1. Importing NumPy

NumPy is a Python library used for numerical computing.

It provides a special data structure called **ndarray (N-dimensional array)** which is faster and more powerful than normal Python lists.

We usually import NumPy with the alias `np`.

```python
import numpy as np
```

## 2. Creating a NumPy Array

A NumPy array is similar to a Python list, but it is optimized for mathematical operations.

Syntax:
np.array()

```python
arr = np.array([1,2,3,4,5])
print(arr)
```

## 3. Checking the Type

Let's verify what type of object NumPy creates.

```python
print(type(arr))
```

## 4. zero Array
Creates an array filled with zeros

```python
zr = np.zeros((3,4))
print(zr)   
```

## 5. Ones Array

Creates an array filled with ones

```python
ones = np.ones((2,3))
print(ones)
```

## 6. Full Array

Creates an array filled with a specific value.

```python
full_arr = np.full((2, 2), 7)
print(full_arr)
```

## 7.Identity Matrix

creates an identity matrix for given dimension .

Very useful in linear algebra and machine learning.

```python
identity = np.eye(3)
print(identity)
```

## 8. arange()

Creates values in a range with a step size.

```python
range_Array = np.arange(0, 10, 2)
print(range_Array)
```

## 9. linspace( )

Creates evenly spaced numbers between two values.

```python
lin_arr = np.linspace(0, 7, 5)
print(lin_arr)
```

## 10. Shape of an Array

Shape tells us how many rows and columns the array has.

```python
print(arr.shape)
```

Meaning of (5,)

This means:

5 elements
1 dimension

## 11. Number of Dimensions

```python
print(arr.ndim)
```

## 12. Total Number of Elements

```python
print(arr.size)
```

## 13. changing data type

Use `astype( )` to convert data types

```python
a1 = np.array([[1,2,3], [4,5,6]])
print(a1.dtype)

a1 = a1.astype('float64')
print(a1.dtype)   
```

# 14. Repeat and Tile


These are useful for repeating values and patterns.

**Repeat**

```python
repeat_Arr = np.repeat([1,2,3], 9)
print(repeat_Arr)  
```

**Tile**

```python
tile_Arr = np.tile([1,2,3], 9)
print(tile_Arr)  
```

## 15. Mathematical Operations

```python
print(arr + 2)
print(arr * 2)
print(arr ** 2)
```

**Meaning**

This is where NumPy becomes magical.

Normal Python lists can’t do proper math like this.

NumPy applies operations to every element.

Example:

[1,2,3,4,5]+2=[3,4,5,6,7]

This is called vectorized operation.

Very important term.

It means operate on the whole array at once.

Faster and cleaner.
